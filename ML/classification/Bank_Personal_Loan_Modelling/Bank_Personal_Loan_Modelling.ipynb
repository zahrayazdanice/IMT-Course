# Normalize the Dataset

```python
from sklearn import preprocessing 
```

```python
df2 = df1.drop("Personal Loan" , axis=1)

scaler = preprocessing.MinMaxScaler(feature_range=(0,1))
norm = scaler.fit_transform(df2)
normDF=pd.DataFrame(norm , columns=['ID','Age','Experience','Income','ZIP Code','Family','CCAvg','Education','Mortgage','Securities Account','CD Account','Online','CreditCard'])

normDF
```

# Creat Modeol :


Logistic Regression

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
```

```python
x = normDF
y = df1['Personal Loan']
```

```python
# balance the data

from imblearn.over_sampling import SMOTE
smote = SMOTE()
X_resampled, y_resampled = smote.fit_resample(x, y)
```

```python
x_train, x_test, y_train , y_test = train_test_split(X_resampled,y_resampled,test_size=0.2,stratify=y_resampled ,random_state=0)


X_resampled
```

```python
logreg = LogisticRegression(solver='liblinear')
logreg.fit(x_train,y_train)
y_pred = logreg.predict(x_test)
```

```python
print("Accuracy",metrics.accuracy_score(y_test ,y_pred))
```

```python
fpr , tpr,_ =metrics.roc_curve(y_test,y_pred)
plt.plot(fpr,tpr,label="data 1")
plt.legend(loc=4)
plt.show()
```

```python
y_pred_proba = logreg.predict_proba(x_test)[::,1]
fpr , tpr , _ = metrics.roc_curve(y_test , y_pred_proba)
plt.plot(fpr,tpr,label='data 1')
plt.legend(loc=4)
plt.show()
```

# K_nearest neighbors ===> KNN


```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
x = pd.DataFrame(X_resampled).values
y = y_resampled.values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x , y,random_state=0 , test_size=0.2)
```

```python
k =1
knnModel = KNeighborsClassifier(k)
knnModel.fit(x_train,y_train.ravel())
y_pred = knnModel.predict(x_test)
```

```python
print("Accuracy : ",metrics.accuracy_score(y_test , y_pred))
```

```python
# "Selecting the optimal value of K"
```

```python
k=20
Acc=np.zeros((k))


for i in range(1,k+1) :
    knnModel = KNeighborsClassifier(n_neighbors=i)
    knnModel.fit(x_train , y_train.ravel())
    y_pred = knnModel.predict(x_test)
    
    Acc[i-1] = metrics.accuracy_score(y_test,y_pred)
    
Acc    
```

```python
training_acc=[]
test_acc=[]

neighbors_setting = range(1,20)

for i in neighbors_setting :
    knnModel = KNeighborsClassifier(n_neighbors = i)
    knnModel.fit(x_train , y_train.ravel())
    training_acc.append(knnModel.score(x_train , y_train))
    test_acc.append(knnModel.score(x_test, y_test))
```

```python
plt.grid(True)
plt.plot(neighbors_setting,training_acc , label='Accuracy of the training set')
plt.plot(neighbors_setting, test_acc , label='Accuracy of the test set')

plt.ylabel('Accuracy')
plt.xlabel('Number of Neighbors')

plt.legend()

plt.show()
```
