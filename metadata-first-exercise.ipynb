```python

# plt.title("Pairwise Feature Analyses" , fontsize=20)
sns.pairplot(df2 , hue="sex" , markers=["o","D"])
```

```python
df2.shape
```

```python
df2.dtypes
```

```python
sns.set_context("notebook", font_scale=3)

g = sns.catplot(
    data=df2,
    kind="bar",
    x="sex",
    y="age",
    col="dx",
    height=20,
    aspect=.3,
)

g.set_titles("dx={col_name}")
g.fig.suptitle("Average Age by Sex Across Diagnosis Types", fontsize=40 , color="purple")
g.fig.subplots_adjust(top=.85)

plt.show()
```

```python
sns.set_context("notebook", font_scale=1)
sns.histplot(data=df2, x="age", bins=20, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.set_context("notebook", font_scale=1)
sns.histplot(data=df2, x="age", hue="sex", bins=20, kde=True)

plt.title("Age Distribution by Sex")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.set_context("notebook", font_scale=1)
sns.boxplot(data=df2, x="sex", y="age")

plt.title("Age Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Age")
plt.show()
```

```python
sns.countplot(data=df2, x="dx")

plt.title("Distribution of Diagnosis Types")
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(
    data=df2,
    y="localization",
    order=df2["localization"].value_counts().index
)

plt.title("Distribution of Lesion Localization")
plt.xlabel("Count")
plt.ylabel("Localization")
plt.show()
```

```python
sns.countplot(data=df2, x="dx", hue="sex")

plt.title("Diagnosis Distribution by Sex")
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=df2, y="localization", hue="sex")

plt.title("Lesion Localization by Sex")
plt.xlabel("Count")
plt.ylabel("Localization")
plt.show()
```

```python
sns.boxplot(data=df2, x="dx", y="age")

plt.title("Age Distribution Across Diagnosis Types")
plt.xlabel("Diagnosis")
plt.ylabel("Age")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(
    data=df2,
    x="dx",
    y="age",
    hue="sex"
)

plt.title("Age Distribution by Diagnosis and Sex")
plt.xlabel("Diagnosis")
plt.ylabel("Age")
plt.xticks(rotation=45)
plt.show()
```
