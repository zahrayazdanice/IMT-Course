```python
for i in df.columns :
    df[i] = df[i].replace(["> 1000"] , '1001')
    
df.isin(["> 1000"]).sum()    
```

```python
for i in df.columns :
    df[i]=df[i].replace(["-"],np.nan)
    
    
df.isin(["-"]).sum()    
```

```python
species = (df_iran ['Institution'])
penguin_means = {
    'Score': df_iran ['Score'],
    'National Rank': df_iran ['National Rank'],
    'World Rank': df_iran ['World Rank'], 
    'Influence':df_iran['Influence'], 
    
}

x = np.arange(len(species))  # the label locations
width = 0.2  # the width of the bars

colors = ['darkcyan', 'indigo', 'yellow', 'red']  # Replace with your desired colors

fig, ax = plt.subplots(figsize=(25, 10), dpi=200)
rects = []
for i, (label, values) in enumerate(penguin_means.items()):
    rects.append(ax.bar(x + i * width, values, width, label=label, color=colors[i]))

ax.set_xlabel('Institution', fontsize=18)
ax.set_ylabel('Values', fontsize=18)
ax.set_yticks (range (0, 1000, 20))
ax.set_title('Comparison of Metrics by Iran Institution', fontsize=20, fontweight='bold')
ax.set_xticks(x + width, species)
ax.set_xticks(x + (len(penguin_means) - 1) * width / 2)  # Adjust the x-axis ticks to be in the middle of the bars
ax.set_xticklabels(species,rotation = 75, fontsize=14)
ax.legend(loc = 'best', fontsize=14)
ax.grid()

plt.show()
```
